```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import kurtosis


```

```python
# Dados
dados = {
    "Ano": [2022, 2023, 2024, 2025, 2026],
    "Hipertensos": [803, 1668, 1202, 4200, 4960],
    "Diabéticos": [443, 744, 750, 2114, 2620],
    "Domiciliados": [142, 289, 540, 1120, 1054],
    "Crianças": [353, 498, 401, 2144, 2522]
}

df = pd.DataFrame(dados)

# Estatísticas descritivas
print("Médias:\n", df.mean(numeric_only=True))
print("\nMedianas:\n", df.median(numeric_only=True))
print("\nModas:\n", df.mode().iloc[0])

# Amplitude
print("\nAmplitude:\n", df.max(numeric_only=True) - df.min(numeric_only=True))

# Variância populacional e amostral
print("\nVariância populacional:\n", df.var(ddof=0, numeric_only=True))
print("\nVariância amostral:\n", df.var(ddof=1, numeric_only=True))

# Desvio padrão populacional e amostral
print("\nDesvio padrão populacional:\n", df.std(ddof=0, numeric_only=True))
print("\nDesvio padrão amostral:\n", df.std(ddof=1, numeric_only=True))

# Quartis e percentis
print("\nQuartis (25%, 50%, 75%):\n", df.quantile([0.25, 0.5, 0.75]))
print("\nPercentil 90:\n", df.quantile(0.9))

# Coeficiente de variação (CV = std / mean)
cv = df.std(ddof=1, numeric_only=True) / df.mean(numeric_only=True)
print("\nCoeficiente de variação:\n", cv)

# Covariância
print("\nCovariância:\n", df.cov())

# Correlação de Pearson
print("\nCorrelação de Pearson:\n", df.corr(method="pearson"))
```

```python
# (a) Lei dos Grandes Números
def lei_grandes_numeros(n_lancamentos=10000):
    # Simulação de moeda justa (0 = cara, 1 = coroa)
    lancamentos = np.random.randint(0, 2, n_lancamentos)
    freq_relativa = np.cumsum(lancamentos) / np.arange(1, n_lancamentos+1)

    plt.figure(figsize=(10,5))
    plt.plot(freq_relativa, label="Frequência relativa de 'coroa'")
    plt.axhline(0.5, color="red", linestyle="--", label="Valor esperado (0.5)")
    plt.title("Lei dos Grandes Números - Moeda justa")
    plt.xlabel("Número de lançamentos")
    plt.ylabel("Frequência relativa")
    plt.legend()
    plt.show()

# -----------------------------
# (b) Teorema Central do Limite
def teorema_central_limite(dataset, n_amostras=1000, tamanho_amostra=30):
    medias = []
    for _ in range(n_amostras):
        amostra = np.random.choice(dataset, tamanho_amostra, replace=True)
        medias.append(np.mean(amostra))

    plt.figure(figsize=(10,5))
    sns.histplot(medias, kde=True, bins=30, color="blue")
    plt.title(f"TCL - Distribuição das médias amostrais (n={tamanho_amostra})")
    plt.xlabel("Média amostral")
    plt.ylabel("Frequência")
    plt.show()

# -----------------------------
# Exemplo de uso:
hipertensos = [803, 1668, 1202, 4200, 4960]

# Lei dos Grandes Números
lei_grandes_numeros(n_lancamentos=5000)

# Teorema Central do Limite
teorema_central_limite(dataset=hipertensos, n_amostras=2000, tamanho_amostra=3)
teorema_central_limite(dataset=hipertensos, n_amostras=2000, tamanho_amostra=30)
```
